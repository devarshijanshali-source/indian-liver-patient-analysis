```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("indian_liver_patient.csv")

print("Dataset shape:", df.shape)
print(df.head())
print(df.info())
```

```python
print("Missing values:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates().copy()

df["Gender"] = df["Gender"].astype("category")

numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].apply(
    pd.to_numeric, errors="coerce"
)

df["Albumin_and_Globulin_Ratio"] = (
    df["Albumin_and_Globulin_Ratio"].fillna(
        df["Albumin_and_Globulin_Ratio"].median()
    )
)

print(df.dtypes)
print("Missing values after cleaning:\n", df.isnull().sum())
```

```python
for col in numeric_cols:
    if col == "Dataset":
        continue

    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    outliers = df[
        (df[col] < q1 - 1.5 * iqr) |
        (df[col] > q3 + 1.5 * iqr)
    ]

    print(col, "outliers:", len(outliers))
```

```python
sns.set_theme(style="whitegrid")

# 1. Disease distribution
sns.countplot(data=df, x="Dataset")
plt.title("Liver Disease Class Distribution")
plt.show()

# 2. Age distribution
sns.histplot(data=df, x="Age", bins=20, kde=True)
plt.title("Age Distribution")
plt.show()

# 3. Gender distribution
sns.countplot(data=df, x="Gender")
plt.title("Gender Distribution")
plt.show()

# 4. Bilirubin outliers
sns.boxplot(data=df, y="Total_Bilirubin")
plt.title("Total Bilirubin Outliers")
plt.show()

# 5. Albumin by disease class
sns.boxplot(data=df, x="Dataset", y="Albumin")
plt.title("Albumin by Disease Class")
plt.show()

# 6. Correlation heatmap
sns.heatmap(
    df.select_dtypes(include=np.number).corr(),
    annot=True, cmap="coolwarm", fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()

# 7. Age vs bilirubin
sns.scatterplot(
    data=df, x="Age", y="Total_Bilirubin",
    hue="Dataset"
)
plt.title("Age vs Total Bilirubin")
plt.show()
```

```python
df.to_csv("cleaned_liver_dataset.csv", index=False)
print("Cleaned dataset saved successfully!")
```
